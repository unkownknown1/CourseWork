# Automated Password Analyzer Demo
This notebook demonstrates the analyzer with **fictional test passwords only**. Do not enter or store a real password in a notebook because notebook inputs and outputs may be saved. Use `python password_analyzer.py` for hidden interactive input.

```python
from password_analyzer import analyze_password

fictional_examples = [
    "password",
    "Example123!Phrase",
    "Mango-River7!Cloud",
]

for example in fictional_examples:
    result = analyze_password(example)
    print({"score": result.score, "rating": result.rating, "suggestions": result.suggestions})
```
